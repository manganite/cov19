````python
import pandas as pd
````

````python
dtf_cases = pd.read_csv("https://raw.githubusercontent.com/CSSEGISandData/COVID-19/master/csse_covid_19_data/csse_covid_19_time_series/time_series_covid19_confirmed_global.csv", sep=",")
````

````python
dtf_cases = dtf_cases.drop(['Province/State', 'Lat', 'Long'], axis=1).groupby("Country/Region").sum()
dtf_cases.columns = pd.to_datetime(dtf_cases.columns, format='%Y-%m-%d', infer_datetime_format=True)
dtf_cases.index.name="Country"

dtf_cases
````

````python
import plotly.graph_objects as go
````

````python
fig = go.Figure(
    data=go.Choropleth(
        locationmode = "country names",
        locations = dtf_cases.index,
        z = dtf_cases[dtf_cases.columns[-1]],
        text = dtf_cases.index,
        colorscale = 'matter',
        reversescale=False,
        autocolorscale=False,
        marker_line_color='darkgray',
        marker_line_width=0.5
    )                                          
)
````

````python
from ipywidgets import interact
max_days = len(dtf_cases.columns)-1
fig_wg = go.FigureWidget(fig)
fig_wg.update_layout(title_text='Cummulated cases worldwide ' + dtf_cases.columns[max_days].strftime('%d/%m/%Y'))
@interact(day=(0, max_days, 1))
def update(day=max_days):
    date = dtf_cases.columns[day]
    fig_wg.update_layout(title_text='Cummulated cases worldwide ' + date.strftime('%d/%m/%Y'))
    with fig_wg.batch_update():
        fig_wg.data[0].z = dtf_cases[date]

fig_wg
````
